## Unit 03: Operators, powers and energy units
### الوحدة 03: العمليات الرياضية ووحدات الطاقة

Use ** for exponentiation, / for division, and * for multiplication. Electrical energy in kWh equals power in kW multiplied by time in hours.

### Line-by-line explanation | شرح كل سطر

1. **L1** — Set the appliance power to 1500 W.
   
   نحدد استطاعة الجهاز بـ 1500 واط.

## Unit 04: Conditions: if, elif and else
### الوحدة 04: الشروط واتخاذ القرار

Indentation determines which instructions belong to a condition. The code chooses an engineering message according to the battery state of charge.

تحدد المسافات البادئة التعليمات التابعة للشرط. يختار المثال رسالة هندسية بحسب حالة شحن البطارية.

**MATLAB → Python:** `MATLAB: if soc < 20 ... elseif soc < 80 ... else ... end`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Record state of charge as 35 percent.
   
   نسجل حالة الشحن بنسبة 35 بالمئة.
2. **L2** — Ask whether charge is below 20 percent; the colon opens a block.
   
   نفحص هل الشحن أقل من 20 بالمئة؛ النقطتان تفتتحان الكتلة.
3. **L3** — Choose the low-battery text only if the condition is true.
   
   نختار نص انخفاض الشحن إذا تحقق الشرط فقط.
4. **L4** — Otherwise check whether the value is below 80 percent.
   
   وإلا نختبر هل النسبة أقل من 80 بالمئة.
5. **L5** — Store the normal-range message for 35 percent.
   
   نحفظ رسالة المجال الطبيعي عند 35 بالمئة.
6. **L6** — Otherwise handle the remaining values.
   
   وإلا نعالج القيم الباقية.
7. **L7** — Choose nearly full for values of at least 80 percent.
   
   نختار «قريب من الامتلاء» عندما تكون النسبة 80 أو أكثر.
8. **L8** — Print the selected message after the conditional block.
   
   نعرض الرسالة المختارة بعد انتهاء الشروط.

```python
soc = 35  # percent
if soc < 20:
    message = 'Low battery'
elif soc < 80:
    message = 'Normal range'
else:
    message = 'Nearly full'
print(message)
```

## Unit 05: Loops: for and while
### الوحدة 05: الحلقات التكرارية

A for loop processes a known series of measurements. A while loop repeats only while its condition remains true; always make sure it can stop.

تعالج حلقة for سلسلة قياسات محددة. وتستمر while ما دام شرطها متحققا؛ يجب ضمان إمكانية توقفها.

**MATLAB → Python:** `MATLAB: for k = 1:length(loads), ... end; Python indices begin at 0.`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Define four hourly power readings in kW.
   
   نحدد أربع قراءات استطاعة ساعية بالكيلوواط.
2. **L2** — Start the accumulated energy at zero.
   
   نبدأ الطاقة المتراكمة من الصفر.
3. **L3** — Process each load value in sequence.
   
   نعالج كل قيمة حمل على التوالي.
4. **L4** — Add load times 1 hour to the cumulative energy.
   
   نضيف استطاعة الحمل مضروبة بساعة واحدة إلى الطاقة.
5. **L5** — Print the total energy: 6 kWh.
   
   نطبع مجموع الطاقة وهو 6 كيلوواط ساعي.
6. **L6** — Initialize the repetition counter at zero.
   
   نبدأ عداد التكرار من الصفر.
7. **L7** — Repeat while the counter is less than two.
   
   نكرر ما دام العداد أصغر من اثنين.
8. **L8** — Print each check number.
   
   نطبع رقم الفحص في كل دورة.
9. **L9** — Increase the counter to guarantee termination.
   
   نزيد العداد لضمان توقف الحلقة.

## Unit 07: Functions and input validation
### الوحدة 07: الدوال والتحقق من المدخلات

A function packages a calculation into a reusable unit. def defines it, parameters receive inputs, and return sends back the result.

تجمع الدالة عملية حسابية يمكن استخدامها مرات متعددة. تعرّف def الدالة، وتستقبل المعاملات المدخلات، ثم تعيد return النتيجة.

**MATLAB → Python:** `MATLAB: function P = dc_power(V,I); P = V*I; end`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Define a function with voltage and current inputs.
   
   نعرّف دالة تستقبل الجهد والتيار.
2. **L2** — Check whether any input is negative.
   
   نتحقق هل أي مدخل سالب.
3. **L3** — Raise a clear error for unsupported inputs.
   
   نظهر رسالة خطأ واضحة للمدخلات غير المدعومة.
4. **L4** — Return the product of voltage and current.
   
   نعيد ناتج ضرب الجهد بالتيار.
5. **L5** — Call the function for 24 V and 3 A.
   
   نستدعي الدالة لجهد 24 فولت وتيار 3 أمبير.
6. **L6** — Call it again for 48 V and 2 A.
   
   نستدعيها مرة أخرى لجهد 48 فولت وتيار 2 أمبير.
7. **L7** — Print both calculated power values.
   
   نطبع قيمتي الاستطاعة المحسوبتين.

```python
def dc_power(voltage_v, current_a):
    if voltage_v < 0 or current_a < 0:
        raise ValueError('Use positive values')
    return voltage_v * current_a
p1 = dc_power(24, 3)
p2 = dc_power(48, 2)
print(p1, p2)
```

**MATLAB → Python:** `MATLAB: power = voltage .* current; Python NumPy: power = voltage * current`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Import NumPy and give it the short name np.
   
   نستورد مكتبة NumPy ونسميها اختصارا np.
2. **L2** — Create an array of three voltage values.
   
   ننشىء مصفوفة تضم ثلاث قيم للجهد.

   نضرب القيم المتقابلة عنصرا بعنصر.
5. **L5** — Calculate the average power.
   
   نحسب متوسط الاستطاعة.
6. **L6** — Limit the voltage values to the range 222–230 V for illustration.
   
   نحصر قيم الجهد بين 222 و230 فولت للتوضيح.

   نحوّل الناتج إلى قائمة ونعرضه.

```python
import numpy as np
voltage = np.array([220, 225, 230])
current = np.array([2, 3, 4])
power = voltage * current
average = np.mean(power)
safe_voltage = np.clip(voltage, 222, 230)
print(power.tolist())
print(round(average, 2))
print(safe_voltage.tolist())
```

## Unit 11: Complex power in AC circuits
### الوحدة 11: الاستطاعة العقدية في دارات التيار المتناوب

Python uses j for the imaginary unit. For RMS phasors and passive sign convention, complex power is S = V × conjugate(I), with real part P and imaginary part Q.

تستخدم Python الحرف j للوحدة التخيلية. عند استعمال الفازورات بالقيم الفعالة واتفاقية الحمل تكون الاستطاعة العقدية S = V × مرافق(I)، وجزؤها الحقيقي P والتخيلي Q.

**MATLAB → Python:** `MATLAB: S = V * conj(I); Python: S = V * I.conjugate()`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Import complex-math functions.
   
   نستورد دوال العمليات على الأعداد العقدية.
2. **L2** — Represent a 230-V RMS reference phasor.
   
   نمثل فازور جهد مرجعي بالقيمة الفعالة 230 فولت.
3. **L3** — Represent 8-A RMS current lagging voltage by 30 degrees.
   
   نمثل فازور تيار فعال 8 أمبير متأخرا عن الجهد بـ 30 درجة.
4. **L4** — Multiply voltage by the current's complex conjugate.
   
   نضرب الجهد بالمرافق العقدي للتيار.
5. **L5** — Extract active power from the real part of S.
   
   نستخرج الاستطاعة الفعالة من الجزء الحقيقي لـ S.
6. **L6** — Extract reactive power from the imaginary part of S.
   
   نستخرج الاستطاعة الردية من الجزء التخيلي لـ S.
7. **L7** — Print approximate active power in watts.
   
   نطبع الاستطاعة الفعالة التقريبية بالواط.
8. **L8** — Print approximate reactive power in var.
   
   نطبع الاستطاعة الردية التقريبية بالفار.

```python
import cmath
voltage = 230 + 0j
current = 8 * cmath.exp(-1j * cmath.pi / 6)
s = voltage * current.conjugate()
p_w = s.real
q_var = s.imag
print(round(p_w), 'W')
print(round(q_var), 'var')
```

## Unit 12: Mini-project: PV, load and battery
### الوحدة 12: المشروع المصغر: الطاقة الشمسية والحمل والبطارية

A first energy-balance simulation with one-hour steps, a 90% one-way charge/discharge efficiency, no export, and no power limit. It teaches algorithms, not a complete real-world design.

محاكاة أولية لميزان الطاقة بخطوة زمنية قدرها ساعة، وكفاءة شحن وتفريغ أحادية الاتجاه 90%، دون تصدير أو حد للاستطاعة. الهدف تعلم الخوارزمية وليس اعتماد تصميم هندسي جاهز.

**MATLAB → Python:** `MATLAB: use a for loop, if/else, and vector indexing 1...N. Python indexes 0...N-1.`

### Line-by-line explanation | شرح كل سطر

1. **L1** — Enter four hourly loads in kW.
   
   ندخل أربعة أحمال ساعية بالكيلوواط.
2. **L2** — Enter PV generation during the same four hours.
   
   ندخل استطاعة الألواح الشمسية خلال الساعات نفسها.
3. **L3** — Set the initial stored battery energy to 2 kWh.
   
   نحدد الطاقة المخزنة أوليا في البطارية بـ 2 كيلوواط ساعي.
4. **L4** — Set the battery's idealized energy capacity to 4 kWh.
   
   نحدد السعة الطاقية للبطارية بـ 4 كيلوواط ساعي.
5. **L5** — Specify a 90% one-way efficiency.
   
   نحدد كفاءة أحادية الاتجاه مقدارها 90 بالمئة.
6. **L6** — Loop over all four one-hour periods.
   
   نكرر الحساب على الفترات الأربع، طول كل منها ساعة.
7. **L7** — Calculate PV generation minus load.
   
   نحسب استطاعة التوليد الشمسي ناقص الحمل.
8. **L8** — If the result is nonnegative, PV covers the load.
   
   إذا كانت النتيجة غير سالبة فالتوليد الشمسي يغطي الحمل.
9. **L9** — Limit charging to available surplus and remaining storage capacity.
   
   نحد الشحن بالفائض المتاح والمساحة المتبقية في البطارية.
10. **L10** — Update stored energy with charging losses.
   
   نحدّث الطاقة المخزنة مع خسائر الشحن.
11. **L11** — No grid import is required while generation meets demand.
   
   لا حاجة للسحب من الشبكة ما دام التوليد يغطي الطلب.
12. **L12** — Otherwise the load exceeds PV generation.
   
   وإلا يكون الحمل أكبر من التوليد الشمسي.
13. **L13** — Limit battery discharge by deficit and available energy.
   
   نحد تفريغ البطارية بالعجز والطاقة المتاحة.
14. **L14** — Subtract battery energy including discharge losses.
   
   نطرح طاقة البطارية مع أخذ خسائر التفريغ بالحسبان.
15. **L15** — Import from grid only the deficit not supplied by the battery.
   
   نسحب من الشبكة فقط العجز الذي لم تغطه البطارية.
16. **L16** — Print hour, remaining stored energy and grid import.
   
   نطبع الساعة والطاقة المتبقية في البطارية والاستطاعة المسحوبة من الشبكة.

```python
load_kw = [2, 3, 2, 1]
pv_kw = [0, 4, 3, 0]
soc = 2.0  # stored kWh
capacity = 4.0
eta = 0.90
for hour in range(len(load_kw)):
    net_kw = pv_kw[hour] - load_kw[hour]
    if net_kw >= 0:
        charge_kw = min(net_kw, (capacity - soc) / eta)
        soc += charge_kw * eta
        grid_import = 0.0
    else:
        discharge_kw = min(-net_kw, soc * eta)
        soc -= discharge_kw / eta
        grid_import = -net_kw - discharge_kw
    print(hour, round(soc, 2), round(grid_import, 2))
```
